## Customer response drafting (Braden's workload)

Run the first cell of this notebook (path + `.env` setup) first, then these two cells in order.

1. **Generate:** sends the 25 enquiries in `data/processed/customer_enquiries.csv` to all 4 models (100 Bedrock calls). Writes `data/outputs/customer_response_drafting_results.csv`.
2. **Judge:** scores only this workload with the LLM judge (100 calls). Writes `data/outputs/customer_response_drafting_evaluation.csv`. It does **not** re-judge the other workloads.

`expected_response_points` and `difficulty` are not sent to the models (excluded in `workload_runner.py`). Estimated total cost: about $0.05.

```python
from pathlib import Path

import pandas as pd
from workload_runner import run_workload

ROOT = Path.cwd().parent
INPUT_FILE = ROOT / 'data' / 'processed' / 'customer_enquiries.csv'

run_workload(workload='customer_response_drafting', input_file=INPUT_FILE)

results = pd.read_csv(ROOT / 'data' / 'outputs' / 'customer_response_drafting_results.csv')
print(results.groupby('model_name')['status'].value_counts())
```

```python
from llm_judge import evaluate_workload

evaluate_workload('customer_response_drafting')

evaluation = pd.read_csv(ROOT / 'data' / 'outputs' / 'customer_response_drafting_evaluation.csv')
print(evaluation.groupby('model_name')['overall_score'].agg(['mean', 'count']))
```
